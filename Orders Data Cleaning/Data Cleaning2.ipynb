```python
min_count = data['Country'].value_counts().min()

balanced_list = []
for city, group in data.groupby('Country'):
    balanced_list.append(group.sample(min_count))
data_balanced = pd.concat(balanced_list).reset_index(drop=True)

```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
data['Country'].value_counts().plot(kind='bar', ax=axes[0])
data_balanced['Country'].value_counts().plot(kind='bar', ax=axes[1])

plt.show()
```
