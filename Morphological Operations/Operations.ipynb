```python
import cv2
import numpy as np

img = cv2.imread('D:\PythonProject1\Intern\Images\WhatsApp Image 2026-06-21 at 9.36.31 PM.jpeg', cv2.IMREAD_GRAYSCALE)

kernel = np.ones((3,3), np.uint8)

opened = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closed = cv2.morphologyEx(opened, cv2.MORPH_CLOSE, kernel)

cleaned = cv2.bilateralFilter(closed, d=5, sigmaColor=5, sigmaSpace=5)

plt.imshow(cleaned,cmap='gray')
plt.show()
```
