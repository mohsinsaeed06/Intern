```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread('D:\PythonProject1\Intern\Images\images.jpg',cv2.IMREAD_GRAYSCALE)

blurr=cv2.GaussianBlur(img,(3,3),0)
l1,binary=cv2.threshold(blurr,127,255,cv2.THRESH_BINARY)
binary
kernel=np.ones((5,5))
opened=cv2.morphologyEx(binary,cv2.MORPH_OPEN,kernel)
cleaned=cv2.morphologyEx(opened,cv2.MORPH_CLOSE,kernel)
```
