```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Orders.csv")
data_clone=data
print(data.isnull().sum())
```

```python
data['Discount']=data['Discount'].fillna(data['Discount'].mode()[0])
print(data.isnull().sum())
```

```python
data['Country'].unique()
type={'Syria':1,         'Saudi Arabia':2, 'United Arab Emirates':3,
                  'USA':4,                'Egypt':5,              'Morocco':6,
               'France':7}
data['Country']=data['Country'].map(type)
data.head()
```

```python
data['City'].unique()
type=[      'homs',     'riyadh',  'abu dhabi', 'washington',     'aleppo',
      'cairo',      'aseer',  'las vegas', 'casablanca',      'asyut',
      'paris',     'jeddah',   'new york',   'damascus',      'dubai',
     'dammam',   'fujairah']
dict={val:i for i,val in enumerate(type)}
print (dict)

data['City']=data['City'].map(dict)
```

```python
data['Category'].unique()
type=[         'Tablet',      'Smartphone',  'Digital Camera',      'Headphones',
      'Smartwatch',          'Laptop',   'Smart Speaker',      'VR Headset',
 'Fitness Tracker',  'Gaming Console']
dict={val:i for i,val in enumerate(type)}
print (dict)

data['Category']=data['Category'].map(dict)
```

```python
min_count = data['Country'].value_counts().min()

balanced_list = []
for city, group in data.groupby('Country'):
    balanced_list.append(group.sample(min_count))
data_balanced = pd.concat(balanced_list).reset_index(drop=True)

```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
data['Country'].value_counts().plot(kind='bar', ax=axes[0])
data_balanced['Country'].value_counts().plot(kind='bar', ax=axes[1])

plt.show()
```
