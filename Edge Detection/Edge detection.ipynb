```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python

image_path = 'D:\PythonProject1\Intern\Edge Detection\image.png' 
image = cv2.imread(image_path)
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
blurred_image = cv2.GaussianBlur(gray_image, (3,3),0)
canny = cv2.Canny(blurred_image,50,150)
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(canny, cmap='gray')



plt.show()
```

```python
image=cv2.imread(image_path)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
blurred_image = cv2.GaussianBlur(gray_image, (3,3),0)
```

```python
sobelx = cv2.Sobel(src=blurred_image, ddepth=cv2.CV_64F, dx=1, dy=1 ) 
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(sobelx, cmap='gray')



plt.show()
```

```python
x=np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
y=np.array([[-1,-1,-1],[0,0,0],[1,1,1]])
```

```python
prewittx=cv2.filter2D(blurred_image,cv2.CV_64F,x)
prewitty=cv2.filter2D(blurred_image,cv2.CV_64F,y)
mag=np.sqrt(prewittx**2+prewitty**2)
```

```python
final=cv2.convertScaleAbs(mag)
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(final, cmap='gray')



plt.show()
```

```python
scharr = cv2.Scharr(src=blurred_image, ddepth=cv2.CV_64F, dx=1, dy=0 ) 
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(scharr, cmap='gray')



plt.show()
```

```python
Lap = cv2.Laplacian(blurred_image,cv2.CV_64F ) 
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(Lap, cmap='gray')



plt.show()
```
